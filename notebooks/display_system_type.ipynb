```python
from malsim import Scenario
from tests.conftest import attack_mapping

scenario_files = list(attack_mapping.keys())
scenarios = [Scenario.load_from_file(scenario_file) for scenario_file in scenario_files]
scenarios
```

```python
from maltoolbox.visualization.graphviz_utils import render_model
for scenario in scenarios:
    model = scenario.model
    render_model(model, f"logs/{scenario.model.name}")
```

```python
from maltoolbox.visualization.draw_io_utils import create_drawio_file_with_images
for scenario in scenarios:
    model = scenario.model
    create_drawio_file_with_images(model, output_filename=f"logs/{scenario.model.name}.drawio")
```
